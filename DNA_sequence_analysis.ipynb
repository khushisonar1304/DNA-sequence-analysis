```python
from Bio.Seq import Seq
import matplotlib.pyplot as plt
```

```python
dna_sequence = Seq("ATGCGTACGTTAGCGATCG")
print(dna_sequence)
```

```python
complement = dna_sequence.complement()

print("Complement:",complement)
```

```python
rna = dna_sequence.transcribe()

print("RNA:",rna)
```

```python
protein = dna_sequence.translate()

print("Protein:",protein)
```

```python
bases = ["A","T","G","C"]

counts = [dna_sequence.count(base) for base in bases]

plt.bar(bases, counts)
plt.xlabel("Nucleotide")
plt.ylabel("Frequency")
plt.title("DNA Nucleotide Composition")
plt.show()
```
