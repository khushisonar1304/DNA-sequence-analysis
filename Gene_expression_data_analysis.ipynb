```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Gene":["BRCA1","TP53","EGFR","MYC","AKT1"],
    "Control":[12,20,15,30,18],
    "Treatment":[25,18,35,40,22]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Expression_Change"] = (
    df["Treatment"] - df["Control"]
)
print(df)
```

```python
df["Fold_Change"] = (
    df["Treatment"]/df["Control"]
)
print(df)
```

```python
df.plot(
    x="Gene",
    y=["Control","Treatment"],
    kind="bar"
)
plt.title("Gene Expression Comparison")
plt.ylabel("Expression level")
plt.xlabel("Genes")
plt.tight_layout()
plt.show()
```

```python
print("Mean control expression:")
print(df["Control"].mean())
print("Mean treatment expression:")
print(df["Treatment"].mean())
```
